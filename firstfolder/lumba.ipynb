```python
# higher order function 1 map , 2  filter , 3 reduce
```

```python
numbers = [1,3,4,6,3,34]
```

```python
def square(x):
    return x*x
```

```python
# map always return map values you can convert that using type casting
```

```python
new = map(square,numbers)
print(new)   # that returns map function / address
```

```python
new = list(map(square,numbers)) # we want convert that map into list
```

```python
print(new)
```

```python
# filter function -- new list or new filterd objects
```

```python
def greater_than_9(x):
    if x > 9:
        return True
    else:
        return False
```

```python
l =[1,23,3,3,5,7,9,12,1,4,2,4,5,5,5555,556,6433,223,]
```

```python
new = list(filter(greater_than_9,l)) # we want to typecast here
```

```python
print(new)
```

```python
# we can use lambda function also
```

```python
new = list(filter(lambda x : x<9,l))
```

```python
print(new)
```
