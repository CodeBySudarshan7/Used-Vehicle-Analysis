```python
def clean_data(new):
    #remove users with missing names
    new['users'] = [user for user in new['users'] if user['name'].strip() ]
    for user in new['users']:
        user['friends'] = list(set(user['friends']))
    new['users'] = [user for user in new['users'] if user['friends'] or user['liked_pages']]
    for user in new['pages']:
        user['pages'] = list(set(user['id']))
    
    return new
   
    
```
