```python
# JSON (JavaScript Object Notation) is a lightweight data format used for data exchange between servers and applications. 
# It is widely used in APIs, web applications, and configurations.
```

```python
import json
```

```python
sudu = {"month":"july","age":"234","religion":"hindu"}
```

```python
p = json.dumps(sudu)
print(type(p))
```

```python
p
```

```python
 d = json.loads('{"month":"july","age":"234","religion":"hindu"}')
```

```python
file = open("data.json","w")
```

```python
with open("data.json", "w") as file:
    json.dump(sudu, file)
```

```python
file.close()
```

```python
with open("data.json", "r") as file:
    python_data = json.load(file)
```
