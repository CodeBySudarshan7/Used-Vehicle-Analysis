```python
class nandu:
    def __init__(self,name,salary,age):
        self.s23 = name  # s23 -- is an instance attribute that assign value of salary we can use taht value in anywhere in the class
        self.s24 = salary
        self.s25 = age
    def get_salary(self):
        return self.s23
    def get_infog(self):
        print(f"my name is {self.s23}, i have earn salary {self.s24}")
```
