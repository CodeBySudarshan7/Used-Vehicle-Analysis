```python
#Inheritance in Python is a fundamental concept of Object-Oriented Programming (OOP) that allows a new class (the child or derived class) to inherit attributes and methods from an existing class (the parent or base class).
#This promotes code reusability and helps establish a hierarchical structure among classes. 
```

```python
class Animal :
    def __init__(self,dog,bark):
        self.p = dog
        self.q = bark
    def get_name(self):
        print(f" {self.p} , {self.q}")
    
```

```python
m = Animal("h","p")
```

```python
m.get_name()
```

```python
class Animal :
    def __init__(self,name):
        self.p = name
    def speak(self):
        print("speaking here ... ")
class dog(Animal) :
    
    def speak(self):
        super().speak()
        print("barkspeed")
        
```

```python
d = dog("great_dane")
```

```python
d.speak()
```

```python
print(d.speak())
```
