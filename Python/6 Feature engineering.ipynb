```python
import pandas as pd
import numpy as np 
from datetime import datetime
```

```python
df["saledate"] = pd.to_datetime(
    df["saledate"],
    errors="coerce",
    utc=True
)
```

```python
# Feature engineering

df["Vehicle Age"] = (
    df["saledate"].dt.year - df["year"]
)

# Remove non-usable ages

df.loc[
    (df["Vehicle Age"] < 0) |
    (df["Vehicle Age"] > 50),
    "Vehicle Age"
] = np.nan
```

```python
df["Vehicle Age"].count()
```

```python
# Price Difference

df["Price Difference"] = (
    df["sellingprice"] - df["mmr"]
)
```

```python
# Price Difference %

df["Premium %"] = np.where(
    df["mmr"] > 0,
    (df["sellingprice"] - df["mmr"]) / df["mmr"],
    np.nan
)
```

```python
# Premium / Discount

df["Pricing Status"] = np.select(
    [
        df["Price Difference"] > 0,
        df["Price Difference"] < 0,
        df["Price Difference"] == 0
    ],
    [
        "Premium",
        "Discount",
        "At MMR"
    ],
    default="Unknown"
)
```

```python
bins = [
    0,
    10000,
    20000,
    30000,
    40000,
    50000,
    60000,
    70000,
    80000,
    90000,
    100000,
    np.inf
]

labels = [
    "0–10K",
    "10K–20K",
    "20K–30K",
    "30K–40K",
    "40K–50K",
    "50K–60K",
    "60K–70K",
    "70K–80K",
    "80K–90K",
    "90K–100K",
    "100K+"
]

df["Mileage Band"] = pd.cut(
    df["odometer"],
    bins=bins,
    labels=labels,
    right=False
)
```
