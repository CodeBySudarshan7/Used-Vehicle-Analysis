```python
# Create a temporary cleaned color column
df["color_clean"] = (
    df["color"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Standard valid colors
valid_colors = {
    "beige",
    "black",
    "blue",
    "brown",
    "burgundy",
    "charcoal",
    "gold",
    "gray",
    "green",
    "lime",
    "off-white",
    "orange",
    "pink",
    "purple",
    "red",
    "silver",
    "turquoise",
    "white",
    "yellow"
}

# Convert em dash to missing
df.loc[df["color_clean"] == "—", "color_clean"] = pd.NA

# Anything not in valid_colors becomes missing
df.loc[
    df["color_clean"].notna() &
    ~df["color_clean"].isin(valid_colors),
    "color_clean"
] = pd.NA
```

```python
df["color_clean"].value_counts(dropna=False)
```
