```python
df.columns = [
    "Year",
    "Make",
    "Model",
    "Trim",
    "Body",
    "Transmission",
    "VIN",
    "State",
    "Condition",
    "Odometer",
    "Color",
    "Interior",
    "Seller",
    "MMR",
    "SellingPrice",
    "SaleDate"
]
```

```python
###   Standardize text case

proper_cols = [
    "Make", "Model", "Trim",
    "Body", "Transmission",
    "Color", "Interior", "Seller"
]

for col in proper_cols:
    df[col] = df[col].str.title()
```

```python
# Standardize text case

df["VIN"] = df["VIN"].str.upper()
```

```python
# Standardize text case

df["State"] = df["State"].str.upper()
```

```python
# Standardize Transmission

df["Transmission"] = (
    df["Transmission"]
    .str.lower()
    .str.strip()
)

df["Transmission"] = np.select(
    [
        df["Transmission"].eq("automatic"),
        df["Transmission"].eq("manual"),
        df["Transmission"].isna() | df["Transmission"].eq("")
    ],
    [
        "Automatic",
        "Manual",
        "Unknown"
    ],
    default="Unknown"
)
```

```python
# Numeric validation

# Year

current_year = pd.Timestamp.today().year

df.loc[
    (df["Year"] < 1980) | (df["Year"] > current_year),
    "Year"
] = np.nan

# Odometer

df.loc[df["Odometer"] <= 0, "Odometer"] = np.nan

# MMR

df.loc[df["MMR"] <= 0, "MMR"] = np.nan

# Selling Price

df.loc[df["SellingPrice"] <= 0, "SellingPrice"] = np.nan

# Condition

df.loc[df["Condition"] <= 0, "Condition"] = np.nan
```
