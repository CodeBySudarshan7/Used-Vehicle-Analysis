```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
# Year Quality

current_year = datetime.now().year

df["Year Quality"] = np.select(
    [
        df["year"].isna(),
        (df["year"] < 1980) | (df["year"] > current_year)
    ],
    [
        "Missing",
        "Invalid"
    ],
    default="Valid"
)
```

```python
# Odometer Quality

df["Odometer Quality"] = np.select(
    [
        df["odometer"].isna(),
        df["odometer"] <= 0
    ],
    [
        "Missing",
        "Invalid"
    ],
    default="Valid"
)
```

```python
# MMR Quality

df["MMR Quality"] = np.select(
    [
        df["mmr"].isna(),
        df["mmr"] <= 0
    ],
    [
        "Missing",
        "Invalid"
    ],
    default="Valid"
)
```

```python
# SellingPrice Quality

df["SellingPrice Quality"] = np.select(
    [
        df["sellingprice"].isna(),
        df["sellingprice"] <= 0
    ],
    [
        "Missing",
        "Invalid"
    ],
    default="Valid"
)
```

```python
# SaleDate Quality

df["SaleDate Quality"] = np.where(
    df["saledate"].isna(),
    "Missing",
    "Valid"
)
```

```python
# VIN Quality

vin_length = df["vin"].fillna("").str.len()

df["VIN Quality"] = np.select(
    [
        df["vin"].isna() | df["vin"].eq(""),
        vin_length < 10
    ],
    [
        "Missing",
        "Invalid"
    ],
    default="Valid"
)
```

```python
# Overall Quality

quality_cols = [
    "Year Quality",
    "Odometer Quality",
    "MMR Quality",
    "SellingPrice Quality",
    "SaleDate Quality",
    "VIN Quality"
]

df["Overall Quality"] = np.select(
    [
        df[quality_cols].eq("Invalid").any(axis=1),
        df[quality_cols].eq("Missing").any(axis=1)
    ],
    [
        "Invalid",
        "Review"
    ],
    default="Valid"
)
```
